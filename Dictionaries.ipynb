### using fromkeys()
**The fromkeys() method creates a dictionary from the given sequence of keys and values.**

```python
alphabets = {'a','b','c'}
numbers = 1
```

```python
dict1 = dict.fromkeys(alphabets,numbers)
dict1
```

```python
l1 = ['a','b','c']
l2 = [[1,2],[3,4]]
dict2 = dict.fromkeys(l1,l2)
dict2
```

```python
l1 = {'a','b','c'}
l2 = [1]
dict3 = dict.fromkeys(l1,l2)
dict3
```

```python
l2.append(2)
dict3
```

**We have not provided any values, so all the keys are assigned None by default.**

```python
l1 = {'a','b','c'}
dict4 = dict.fromkeys(l1)
dict4
```

## Updating Values in Dictionary

```python
country_capitals = {
  "Germany": "Berlin", 
  "Italy": "Naples", 
  "England": "London"
}
```

### We can also use Update() method

**The update() method updates the dictionary with the elements from another dictionary object or from an iterable of key/value pairs.(ex: Tuples)**

```python
marks = {'Phyiscs' : 70, 'Maths' : 80}
internal_marks = {'Practicals' : 35}
marks.update(internal_marks)
marks
```

```python
d = {'one' : 1, 'two' : 3}
d1 = {'two' : 2}
d.update(d1)
d
```

**Note: The update() method adds element(s) to the dictionary if the key is not in the dictionary. If the key is in the dictionary, it updates the key with the new value.**

```python
d2 = {'three' : 3}
d.update(d2)
d
```

**The update() method also takes an iterable object of key/value pairs (generally tuples).**

```python
letters_dict = {'a' :1}
letters_dict.update([('b',2),('c',3)])
letters_dict
```

## Copy Dictionary

### copy() method returns a shallow copy of the dictionary. It doesn't modify the original dictionary.

```python
original = {'one':1, 'two':2}
new = original.copy()
print(f'new : {new}')
print(f'original : {original}')
```

***Dictionary copy() Method Vs = Operator\
 When the copy() method is used, a new dictionary is created which is filled with a copy of the references from the original dictionary.\
 But When the = operator is used, a new reference to the original dictionary is created.***

```python
new = original
new
```

```python
new.clear()
```

### Here, when the new dictionary is cleared, the original dictionary is also cleared.

```python
print(f'new : {new}')
print(f'original : {original}')
```

```python
original = {'one':1, 'two':2}
new = original.copy()
new.clear()
```

### Here, when the new dictionary is cleared, the original dictionary remains unchanged.

```python
print(f'new : {new}')
print(f'original : {original}')
```
